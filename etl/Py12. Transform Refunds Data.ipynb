2 ways to extract substring from string, based on condition:
1) Split Function
2) RegExp

```python
df = spark.table('gizmobox.bronze.py_refunds')
display(df)
```

```python
from pyspark.sql import functions as F
df_final = df.select(
    "refund_id",
    "payment_id",
    F.date_format("refund_timestamp", "yyyy-MM-dd").alias("refund_date"),
    F.date_format("refund_timestamp", "HH:mm:ss").alias("refund_time"),
    "refund_amount",
    F.regexp_extract("refund_reason", '^([^:]+):', 1).alias("refund_reason"),
    F.regexp_extract("refund_reason", '^[^:]+:(.*)$', 1).alias("refund_source"),
)

display(df_final)
```

```python
df_final.writeTo('gizmobox.silver.py_refunds').createOrReplace()
```
