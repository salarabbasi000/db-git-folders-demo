Select Metadata (for Audit purpose)

```python
%sql
SELECT input_file_name() as filePath, --this function is deprecated for DBX runtime >= 13.3
_metadata.file_path as filePath,
* FROM json.`dbfs:/Volumes/gizmobox/landing/operational_data/customers/`
```

Create View

syntax:

create view if not exists catalogname.schemaname.viewname

as

select * from EXTERNAL LOC

OR


use catalog catalogname;

use schema schemaname;

create view if not exists viewname
as
select * from EXTERNAL LOC

```python
%sql
create view if not exists gizmobox.bronze.v_customers
as
select
*, _metadata.file_path as file_path
FROM json.`dbfs:/Volumes/gizmobox/landing/operational_data/customers/`
```

```python
%sql
select * from gizmobox.bronze.v_customers;
```
