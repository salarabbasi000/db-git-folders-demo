List Files from Payments Folder

```python
schema = 'payment_id integer, order_id integer, payment_timestamp timestamp, payment_status integer, payment_method string'
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, TimestampType
payments_schema = StructType([
    StructField('payment_id', IntegerType(), True), #3rd option is nullable or not (optional) 
    StructField('order_id', IntegerType(), True),
    StructField('payment_timestamp', TimestampType(), True),
    StructField('payment_status', IntegerType(), True),
    StructField('payment_method', StringType(), True)])
```

```python
df = (
    spark.read
    .format('csv')
    .option('delimiter', ',')
    .schema(payments_schema) #schema or payments_schema, both work. However payments_schema is preferred.
    .load('abfss://gizmobox@salarcourseextdl.dfs.core.windows.net/landing/external_data/payments/')
)
display(df)
```

```python
df.writeTo('gizmobox.bronze.py_payments').createOrReplace()
```

```python
%sql
select * from gizmobox.bronze.py_payments;
```

Create External Table (SQL)

```python
%sql
create table if not exists gizmobox.bronze.payments
(
  payment_id integer,
  order_id integer,
  payment_timestamp timestamp,
  payment_status integer,
  payment_method string
) using csv
options (
  header='true',
  delimiter=','
)
location'abfss://gizmobox@salarcourseextdl.dfs.core.windows.net/landing/external_data/payments/'
```

```python
%sql
select * from gizmobox.bronze.payments;
```
