**Extract data from Addresses Files (TSV)**

```python
%fs ls 'dbfs:/Volumes/gizmobox/landing/operational_data/addresses/'
```

Solution: read_files() in SQL. Options in Pyspark

```python
%sql
select * from read_files('dbfs:/Volumes/gizmobox/landing/operational_data/addresses',
  format=>'csv',
  delimiter=>'\t')
```

```python
df = (spark.read
.format("csv")
.option("header", True)
.option("delimiter", '\t')
.load("dbfs:/Volumes/gizmobox/landing/operational_data/addresses")
)
display(df)
```

```python
df.writeTo("gizmobox.bronze.py_addresses").createOrReplace()
```
