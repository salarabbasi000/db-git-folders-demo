Note:
1) for arrays object, accessing it is columnname:attribute.object
2) for json object, accessing it is columnname.attribute.object

Functions:
1) array_distinct(array)
2) explode(array)

```python
%sql
select * from gizmobox.silver.orders_json;
```

```python
%sql
select
json_value.order_id,
json_value.order_status,
json_value.payment_method,
json_value.total_amount,
json_value.transaction_timestamp,
json_value.customer_id,
json_value.items -- need to apply array_distinct() to remove dupes from arrays and explode() to flatten
from gizmobox.silver.orders_json;
```

```python
%sql
select
json_value.order_id,
json_value.order_status,
json_value.payment_method,
json_value.total_amount,
json_value.transaction_timestamp,
json_value.customer_id,
explode(array_distinct(json_value.items))
from gizmobox.silver.orders_json;
```

```python
%sql
create or replace temporary view tv_orders_exploded
as
select
json_value.order_id,
json_value.order_status,
json_value.payment_method,
json_value.total_amount,
json_value.transaction_timestamp,
json_value.customer_id,
explode(array_distinct(json_value.items)) as item
from gizmobox.silver.orders_json;
```

```python
%sql
create table if not exists gizmobox.silver.orders
select
order_id,
order_status,
payment_method,
total_amount,
transaction_timestamp,
customer_id,
item.item_id,
item.name,
item.price,
item.quantity,
item.category,
item.details.brand,
item.details.color
from tv_orders_exploded;
```
