```python
%sql
use catalog gizmobox;

Create schema if not exists landing
	Comment 'landing schema for gizmobox catalog'
	managed location 'abfss://gizmobox@salarcourseextdl.dfs.core.windows.net/landing';

Create schema if not exists gold
	Comment 'gold schema for gizmobox catalog'
	managed location 'abfss://gizmobox@salarcourseextdl.dfs.core.windows.net/gold';

Create schema if not exists silver
	Comment 'silver schema for gizmobox catalog'
	managed location 'abfss://gizmobox@salarcourseextdl.dfs.core.windows.net/silver';

Create schema if not exists bronze
	Comment 'bronze schema for gizmobox catalog'
	managed location 'abfss://gizmobox@salarcourseextdl.dfs.core.windows.net/bronze';
```
