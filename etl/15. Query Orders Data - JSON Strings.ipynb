Top-level object values in databricks: column_name:extraction_path (for eg: value:order_id)

For array elements in objects, use indexes: column_name:extraction_path[index] (for eg: value:items[0])

For leaf-node in objects, use ".": column_name:extraction_path[index].object (for eg: value:items[0].name)

```python
%sql
select 
value:order_id,
value:items[0].item_id::integer,
value:items[1].item_id::integer,
value:items[0],
value:items[1],
value 
from gizmobox.bronze.v_orders;
```
