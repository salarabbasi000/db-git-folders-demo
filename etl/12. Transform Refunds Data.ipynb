2 ways to extract substring from string, based on condition:
1) Split Function
2) RegExp

```python
%sql
create schema if not exists hive_metastore.silver;
```

```python
%sql
create table hive_metastore.silver.refunds
as
select
refund_id,
payment_id,
cast(date_format(refund_timestamp,'yyyy-MM-dd') as date) as refund_date,
date_format(refund_timestamp,'HH:mm:ss') as refund_time,
refund_amount,
regexp_extract(refund_reason,'^([^:]+):') as refund_reason, -- also split(refund_reason,':')[0]
regexp_extract(refund_reason,'^[^:]+:(.*)$') as refund_source -- also split(refund_reason,':')[1]
from
hive_metastore.bronze.refunds
```
