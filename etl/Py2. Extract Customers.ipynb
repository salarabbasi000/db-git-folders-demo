**Extract data from Customers Files (JSON)**

1. Dataframe Reader API: df.read.format(formatdefined).load(path)
2. Or: Dataframe Reader API: df.read.SomeFormatName(path)
3. Dataframe Writer API: df.write.format(formatdefined).mode(modedefined).saveAsTable(tablename)
4. Dataframe Writer API V2: df.writeTo(tablename).SomeV2ModeFunctionName()

Query Single JSON File

```python
%sql
SELECT * FROM json.`dbfs:/Volumes/gizmobox/landing/operational_data/customers/customers_2024_10.json`
-- SELECT * FROM json.`/Volumes/gizmobox/landing/operational_data/customers/customers_2024_10.json (without DFBS)
```

```python
df = spark.read.format('json').load('/Volumes/gizmobox/landing/operational_data/customers/customers_2024_10.json')
display(df)
```

```python
#another way:
df = spark.read.json('/Volumes/gizmobox/landing/operational_data/customers/customers_2024_10.json')
display(df)
```

Query a Folder

```python
%sql
SELECT * FROM json.`dbfs:/Volumes/gizmobox/landing/operational_data/customers/`
```

```python
df = spark.read.format('json').load('/Volumes/gizmobox/landing/operational_data/customers/')
display(df)
```

Select Metadata (for Audit purpose)

```python
%sql
SELECT input_file_name() as filePath, --this function is deprecated for DBX runtime >= 13.3
_metadata.file_path as filePath,
* FROM json.`dbfs:/Volumes/gizmobox/landing/operational_data/customers/`
```

```python
df_with_metadata = df.select("_metadata.file_path","*")
display(df_with_metadata)
```

Writing:

```python
df_with_metadata.write.format('delta').mode('overwrite').saveAsTable('gizmobox.bronze.py_customers')
```

```python
df_with_metadata.writeTo('gizmobox.bronze.py_customers').createOrReplace()
```
