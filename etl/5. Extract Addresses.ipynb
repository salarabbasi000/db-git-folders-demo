**Extract data from Addresses Files (TSV)**

```python
%fs ls 'dbfs:/Volumes/gizmobox/landing/operational_data/addresses/'
```

```python
%sql
select * from csv.`dbfs:/Volumes/gizmobox/landing/operational_data/addresses`
-- These are TSV files. SQL queries doesn't support TSV format. Hence we don't get desired outcome.
```

Solution: read_files() function

```python
%sql
select * from read_files('dbfs:/Volumes/gizmobox/landing/operational_data/addresses',
  format=>'csv',
  delimiter=>'\t')
```

Create View

```python
%sql
create or replace view gizmobox.bronze.v_addresses
as
select * from read_files('dbfs:/Volumes/gizmobox/landing/operational_data/addresses',
  format=>'csv',
  delimiter=>'\t')
```

```python
%sql
select * from gizmobox.bronze.v_addresses;
```
