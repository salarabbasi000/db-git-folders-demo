```python
df = spark.table('gizmobox.bronze.py_memberships')
display(df)
```

```python
from pyspark.sql import functions as F

df_final = df.select(
    F.col('content').alias('membership_card'),
    F.regexp_extract('path','.*/([0-9]+)\\.png$',1).alias('customer_id'),
          )

display(df_final)
```

```python
df_final.writeTo('gizmobox.silver.py_memberships').createOrReplace()
```
