use of pivot for converting the records for shipping & billing addresses into the respective columns.

Pivot: Rows to Cols

```python
df = spark.table('gizmobox.bronze.py_addresses')
display(df)
```

```python
from pyspark.sql import functions as F

df_final = (
    df.groupBy("customer_id")\
    .pivot("address_type",["billing","shipping"])\
    .agg(
        F.max("address_line_1").alias("address_line_1"),
        F.max("city").alias("city"),
        F.max("state").alias("state"),
        F.max("postcode").alias("postcode")
    )
)
display(df_final)
```

```python
df_final.writeTo('gizmobox.silver.py_addresses').createOrReplace()
```
